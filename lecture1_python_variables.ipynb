## Python as a calculator

At its most basic Python can be used as a calculator, performing basic mathematical operations, such as:

```python
4+8
217.81 * 3
18 * ( 29182 - 9382)
-305.8 / 21.1
```

etc. Note that the spacing between the numbers and the operators does not matter to the Python interpreter; however, it is important to make the code readable to humans, so one should try to be consistent in form.

Mathematical order of operations is maintained within the context of evaluating expressions within parentheses first, and also evaluating multiplication and division before addition and subtraction. However, the Python interpreter may not necessarily evaluate a set of multiplications/divisions from left to right. We will see why this might matter shortly.

Python comes with many basic operations that we might expect: +, -, /, *, as well as parentheses (...). Additional operators include:

raising to a power with **, so that 5**2 raises squares five, returning 25; fractional and negative powers may be used
taking the modulus (i.e., finding the remainder of division) with %, so that 5 % 3 returns the remainder of dividing five by three, returning 2.
Think about the results of each of the following, and then evaluate in Python:

```python
3**2
-3**2
10**5
9**0.5
4**(-0.5)
5 % 2
25 % 5
5 % 25
```

Consider the result of simple division: 2.0/3.0. What is the result? Well, in mathematics it would be 0.66666666… with an infinite number of decimal places. However, the computer has finite disk space and finite memory. In general, we cannot expect the computer to have an exact representation of such numbers, nor of irrational $(\sqrt{2}, 7^{1/3})$ and transcendental numbers ($\pi$, Euler’s e, etc.).

## Assignment and variables

The calculator functionality described above is useful at times, but it is not really programming, since we cannot combine results of different calculations and build on previous results, as values aren’t saved.

We now take a large step toward real programming by introducing Python’s assignment operator “=”. Thus, with:

```python
# we will have the result of a calculation (12) stored, which we can access any time later by referring to x. For example:
x = 4 + 8
```

```python
x = 4 + 8
y = 10**3
z = x*x - 10*y
```

The “=” in this context does not express mathematical equivalence (a passive statement that is true or false, just expressing equality). We cannot treat it like an alebraic expression and try rearrange the left hand side (LHS) and right hand side (RHS) to be something like:

```python
x - 4 = 8
```

or anything like that (try it, Python will produce an error).

This operator “=” is an action, providing a reference label to a value that is stored on the computer. In line with this concept, one can read it more like “x is defined to be …”, or “x is set to ….”, or “Let x be …”, etc. As such, every quantity on the RHS must be known and have a value already. Therefore, if c has not been assigned a value previously, then:

```python
d = c*c - 10
c = 4 + 8
# will produce an error when evaluating the first line. Consider the expression:
```

This is not a general relation, like the formula for a straight line in mathematics. This is a single instance, and a specific value for “x” must have been defined previously. (The way to write a general relation like the equation of a line is done by defining a function, discussed later).

There is a careful ordering to how Python treats lines with the assignment operator:

evaluate the RHS (if any term there has not been previously defined, an error occurs)
store the resulting value at some location on the computer (in an appropriate-sized space, based on the final type)
take the label the LHS and associate it with that location, so whenever that label occurs in an expression, Python will go the correct location and get the desired value.
The variable on the LHS of the assignment operator is, well, variable, and its value (and/or type, in Python) can be changed simply by assigning it to a new value at some point. So, at the end of these lines:

```python
x = 10
x = -11.4
x = False
```

there is still only one variable x, and it refers to a quantity of type “bool” and value “False”.

The Python command _whos_ is a useful way to see what variables have been defined in a Python environment. It provides a list of the variable names, their values, their type and other properties.

## Variable names

There are some rules for variable names to keep in mind:

1. They must be made up only of alphanumeric characters (a-z, A-Z, 0-9) and the underscore character _.
2. The variable cannot start with a number, but the number may appear anywhere else.
3. Unusable characters: Special character such as \, $, %, !, etc. cannot* be used, nor can commas ,, periods . or dashes -, as those all have special other meanings. Spaces `` `` cannot be used, as those separate different words.
4. A variable name cannot be a reserved word or a key word (such as for, if, def, etc.). It also should not be word with special meaning (int, True, type, etc.) because the functionality of those original meanings should be maintained.
5. Variables are case sensitive (indeed, all words in Python are). That is, Python distinguishes between upper and lower case, so that Age, age and aGe are three different variables.
Please do remember to make meaningful variable names. For example, we emphasize here translating mathematics/physics/etc. from paper calculations to programs. Therefore, you should employ the same variable names and notations that you have written down to the program you are writing. For example, if writing the formula “distance equals velocity times time”, you should probably make use of the standard variables:

```python
d = v * t
# or perhaps more interpretable, but succinct, names:
dist = vel * time
```

Half (or more!) of the art of programming is to make the code itself as readable and understandable as possible. This makes the program simpler to check by eye while writing and more directly editable at future times. It also greatly reduces the numbers of bugs or errors in the code.

Classify: 
1. <18.5 → "Underweight"
2. 18.5–24.9 → "Normal"
3. 25–29.9 → "Overweight"
4. ≥30 → "Obese"


```python
weight = float(input("Enter your weight in kg: "))
height = float(input("Enter your height in m: "))

bmi = weight / (height ** 2)

if bmi < 18.5:
    print("Underweight")
elif bmi <= 24.9:
    print("Normal")
elif bmi <= 29.9:
    print("Overweight")
else:
    print("Obese")
```

## Reading Data from APIs using Request library

```python
##Reading the IMF data using macraeconomic indicators
import requests
import pandas as pd

# Step 1: Define the base API link
base = "https://www.imf.org/external/datamapper/api/v1"

# Step 2: Choose the indicator ID — here NGDPD means nominal GDP in USD
indicator_id = "PPPPC"

# Step 3: Choose the countries you want
country_ids = ["USA"]

# Step 4: Choose the years you want
years = [2019, 2020, 2021, 2022, 2023, 2024, 2025, 2026, 2027, 2028]

# Step 5: Build the API path (endpoint URL)
path = f"{base}/{indicator_id}"

# Step 6: If you have countries, add them to the path
if country_ids:
    path += "/" + ",".join(country_ids)

# Step 7: If you have years, add them as query parameters
if years:
    years_param = "?periods=" + ",".join(str(y) for y in years)
    path += years_param

# Print the final API URL to see what you are requesting
print("API URL:", path)

# Step 8: Send a GET request to the IMF API
resp = requests.get(path)

# Step 9: Check if the request worked (raises error if not 200 OK)
resp.raise_for_status()

# Step 10: Convert the response from JSON text to a Python dictionary
obj = resp.json()

# Step 11: Extract the values of our indicator
# The JSON structure is: obj["values"][indicator_id] = {country: {year: value}}
values = obj.get("values", {}).get(indicator_id, {})

# Step 12: Convert nested dictionary to a DataFrame
df = pd.DataFrame.from_dict(values, orient="index").T # transpose so years are rows

# Step 13: Convert string numbers to actual numeric values
df = df.apply(pd.to_numeric, errors="coerce")

# Step 14: Display the resulting DataFrame
print(df)
```

## Data Cleaning with Pandas 24th OCTOBER, 2025

```python
## Importing libraries and reading the data
import pandas as pd
import numpy as np
finance = pd.read_csv("finance_data.csv")

#Displaying the top 20 rows
finance.head(6)
```

```python
## Adding Year, Month, Quarter, Week based on date column
finance["Date"] = pd.to_datetime(finance["Date"])

#Adding year column based on date
finance["Year"] = finance["Date"].dt.year

#Adding the quarter column
finance["Quarter"] = finance["Date"].dt.quarter

# Adding month column
finance["Month"] =  finance["Date"].dt.month

finance.head(6)
```

```python
##Plotting line graph of age vs Quarter
finance.groupby("Quarter")["Age"].mean().plot(kind = 'bar')
plt.show()
```

```python
## Converting the age column into positive
finance["Age"] = finance["Age"].abs()
```

```python
## Using BOX plot for AGE
finance["Age"].plot(kind = "box")
```

```python
## Dealing with outliers on Age column/feature/Attribute
## Using IQR method
q1 = finance["Age"].quantile(0.25)
q3 = finance["Age"].quantile(0.75)
IQR = q3 - q1
IQR
```

```python
#Upper bound
up = q3 + 0.5*IQR
#Lower bound
low = q1 - 0.5 *IQR
```

```python
## Remove outliers
finance_clean = finance[(finance["Age"] > low) & (finance["Age"] < up)]
finance_clean["Age"].describe()
```

```python
## Clean numeric column Salary and Bonus
finance_clean["Salary"] = pd.to_numeric(finance_clean["Salary"].replace(r"[^\d.]",
                                                                        "", regex= True), errors= 'coerce')

finance_clean["Bonus"] = pd.to_numeric(finance_clean["Bonus"].replace(r"[^\d.]",
                                                                        "", regex= True), errors= 'coerce')
finance_clean.head(10)
```

```python
## Replace spaces from Name and Department by np.nan
finance_clean["Name"] = finance_clean["Name"].replace(' ', np.nan)
finance_clean["Department"] = finance_clean["Department"].replace(' ', np.nan)
finance_clean["Department"] = finance_clean["Department"].replace('none', np.nan)
finance_clean.head(5)
```

```python
## Inspecting the unique values under Name and Department
finance_clean["Department"].unique()
```

## Data Transformation Scaling, Standadization, Normalization October 29th, 2025
1. Log Transformation helps reduce skewness and compresses the range of values.
2. Z-Score Normalization standardizes the data so it has a mean of 0 and a standard deviation of 1.

Data scaling is essential in machine learning because it ensures that all features contribute equally to the model. Many algorithms, such as those based on distance metrics (e.g., KNN, K-Means) or gradient descent (e.g., linear regression, neural networks), are sensitive to feature magnitude. Without scaling, features with larger ranges dominate, leading to biased results and slower convergence. Scaling also improves optimization stability and makes regularization techniques like Lasso and Ridge effective. Common methods include standardization (Z-score), which centers data around zero with unit variance, and min-max scaling, which normalizes values to a fixed range, typically $[0, 1]$.
$$z = \frac{x - \mu}{\sigma}$$
Where $\mu$ is the Mean of the data points, $\sigma$ is the standard deviation of the dataset, and $x$ is the point

```python
##Numerical column of Age
import pandas as pd

data = {"Age": [20, 25, 30, 15],
       "Name": ["A", "G", "F", "W"]}
df = pd.DataFrame(data)
df
```

```python
## Z-score to transform the Age column
df["age_trans_zsore"] = (df["Age"] - df["Age"].mean())/ df["Age"].std()
df
```

```python
##Scalling using Logarthmic 
import numpy as np
df["age_trans_log"] = np.log(df["Age"])
df
```

```python
#Scaling using MIN-MAX Method
df["age_trans_minmax"] = (df["Age"] - df["Age"].min())/(df["Age"].max() - df["Age"].min())
df
```
