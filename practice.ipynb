# **Penguin Data**

## **What is this about?**

This is data about **penguins** - those black and white birds that live in cold places like Antarctica. They can't fly, but they're excellent swimmers!

Scientists measured 344 penguins and wrote down their:
- **Type** (what kind of penguin)
- **Where they live** (which island)
- **Body measurements**
- **Weight**
- **Male or female**

---

## **What are we measuring?**

Let's look at a penguin's body parts:

```
      /¯¯¯¯¯¯¯¯¯\     ← Bill (beak) length
     /           \ 
    /_____________\
    |             |   ← Bill (beak) depth
```

**Bill (Beak) Measurements:**
- **Length**: How long the beak is
- **Depth**: How tall or thick the beak is

**Other Measurements:**
- **Flipper length**: How long their swimming wings are
- **Body mass**: How much they weigh
- **Sex**: Whether they're male or female

---

## **What can we discover?**

### **1. Compare Penguin Types**
We have 3 types of penguins:
- **Adelie** 
- **Chinstrap** (has a thin black line under its chin)
- **Gentoo** (has a white stripe on its head)

**We can ask:** Are some types bigger than others? Do they have different shaped beaks?

### **2. Male vs Female**
**We can ask:** Are male penguins usually bigger than females? Do they weigh more?

### **3. Where they live**
**We can ask:** Do certain penguin types only live on certain islands?

### **4. Relationships between measurements**
**We can ask:** Do penguins with longer flippers also weigh more? Is there a pattern?

---

## **Why is this useful?**

Even if you don't know anything about penguins, you can be a **data detective**! You can:

- Look for patterns and surprises
- Answer simple questions about size differences
- Make charts to show what you find
- Practice working with real numbers

````python
##Numerical column of Age
import pandas as pd

data = {"Age": [20, 25, 30, 15],
       "Name": ["A", "G", "F", "W"]}
df = pd.DataFrame(data)
df
````

````python
# Z-score

df['age-trans_zscore'] = (df['Age'] - df['Age'].mean()) / df['Age'].std()
df
````

````python
import numpy as np
df['age_trans_log'] = np.log(df['Age'])
df
````

````python
# age tranformation using Min-Max scaling Method

df['age_trans_minmax'] = (df['Age'] - df['Age'].min())  / (df['Age'].max() - df['Age'].min())
df
````
