##  1. What Is the Chi-Square Test?

The **Chi-Square (χ²) Test** is a **statistical test used to examine the relationship between two categorical variables** — or to check whether **observed data fits expected proportions**.

It helps answer questions like:

> “Is there a relationship between gender and loan approval?”
> “Do people in different districts have the same preference for renewable energy?”
> “Is customer satisfaction independent of the type of product purchased?”

---

##  2. Core Idea

Chi-square measures **how different** your **observed counts** (what actually happened) are from **expected counts** (what you’d expect if there were no relationship).

If the difference is large → the variables are **probably related**.
If the difference is small → the variables are **probably independent**.

---

##  3. Two Main Types of Chi-Square Tests

| Type                                | Purpose                                                               | Example                                                                    |
| ----------------------------------- | --------------------------------------------------------------------- | -------------------------------------------------------------------------- |
| **Chi-Square Test of Independence** | Tests if two categorical variables are related                        | Is there an association between *education level* and *employment status*? |
| **Chi-Square Goodness of Fit Test** | Tests if a single categorical variable follows a certain distribution | Are customer choices equally distributed among 4 product brands?           |

## 🇷🇼 4. Real-World Context — Rwanda Rural Support Example

###  Scenario

The **Rwanda Government** supports rural families through a **financial literacy program**.
After one year, researchers want to check whether **the literacy training is linked to savings behavior**.

They collect data:

| Saving Habit            | Received Support | Did Not Receive Support |
| ----------------------- | ---------------- | ----------------------- |
| Saves Regularly         | 120              | 80                      |
| Does Not Save Regularly | 30               | 70                      |



##  5. Step-by-Step Logic (Under the Hood)

1. **State the Hypotheses**

   * **H₀ (Null Hypothesis):**
     There is **no relationship** between receiving support and saving habit.
     (They are *independent*.)

   * **H₁ (Alternative Hypothesis):**
     There **is a relationship** between receiving support and saving habit.
     (They are *dependent*.)

2. **Compute Expected Frequencies**

   Expected value for each cell assumes *no relationship*:

   $$
   E = \frac{(\text{Row Total}) \times (\text{Column Total})}{\text{Grand Total}}
   $$

   For example, for “Saves Regularly & Received Support”:

   $$
   E = \frac{(150 \times 200)}{300} = 100
   $$

3. **Compute the Chi-Square Statistic**

   $$
   \chi^2 = \sum \frac{(O - E)^2}{E}
   $$
   where ( O ) = observed count and ( E ) = expected count.

4. **Compare with Critical Value / Get p-value**

   * Degrees of freedom (df) = (rows - 1) × (columns - 1)
   * If **p < 0.05**, reject ( H₀ )



##  6. Python Example

```python
import pandas as pd
from scipy.stats import chi2_contingency

data = [[120, 80],
        [30, 70]]

chi2, p, dof, expected = chi2_contingency(data)

print("Chi-square Statistic:", chi2)
print("p-value:", p)
print("Degrees of Freedom:", dof)
print("Expected Frequencies:\n", expected)
```

---

##  7. Output Interpretation

| Output       | Meaning                                                          |
| ------------ | ---------------------------------------------------------------- |
| **χ² value** | How large the difference is between observed and expected counts |
| **p-value**  | Probability that this difference happened by chance              |
| **df**       | Number of categories that can vary independently                 |
| **expected** | What counts would look like if variables were independent        |

If p-value < 0.05 → we reject the null hypothesis.
 **Conclusion:** There is a significant association between *support program* and *saving behavior.*

##  9. Key Takeaways

| Concept                                                   | Explanation                                                 |
| --------------------------------------------------------- | ----------------------------------------------------------- |
| **Chi-Square = difference between observed and expected** | Measures deviation from independence                        |
| **Large χ² → small p-value → relationship exists**        | The bigger the difference, the less likely it’s random      |
| **Use only categorical data**                             | Variables must be in categories (e.g., Yes/No, Male/Female) |
| **Don’t use with small counts (<5)**                      | Test becomes unreliable                                     |

---

##  10. Final Intuition

Think of the Chi-Square test as a **truth detector for categorical relationships**.
It tells us:

> “Is what we’re observing truly different — or just random noise?”

When used wisely, it helps decision-makers move from **guessing relationships** → to **proving them with evidence**.
