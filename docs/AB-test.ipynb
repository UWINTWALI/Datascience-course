# **A/B Testing Tutorial — A Data Scientist’s Guide**


## **1. Context: What is A/B Testing?**

A/B testing is a **controlled experiment** that compares **two versions (A and B)** of something to determine **which one performs better**.

It’s used in:

* Product design (buttons, features, layouts)
* Marketing (email versions, ad designs)
* Public policy (program variations)
* Education and health interventions


## **2. The Real-World Example**

### **Scenario:**

A mobile banking startup in Kigali wants to improve **user savings behavior**.

They test **two versions of an SMS message** sent to users every Friday:

* **Version A (Control):** “Save today for a better tomorrow.”
* **Version B (Treatment):** “Your future starts with saving now — deposit today!”

The company wants to know:

> Which message leads to **higher average savings deposits** by the end of the month?


## **3. Experimental Design**

| Component                | Description                                             |
| ------------------------ | ------------------------------------------------------- |
| **Population**           | 2,000 mobile banking users                              |
| **Random Assignment**    | 1,000 users get Message A, 1,000 get Message B          |
| **Measurement Variable** | Monthly deposit amount (in RWF)                         |
| **Goal**                 | See if Message B leads to significantly higher deposits |

---

## **4. Hypotheses**

* **Null hypothesis $(H_0)$**:
  The two messages produce the **same average savings**.
  $$
  \mu_A = \mu_B
  $$

* **Alternative hypothesis $(H_a)$**:
  Message B increases savings.
  $$
  \mu_B > \mu_A
  $$



## **5. Code Example (Python)**

```python
from scipy.stats import ttest_ind
import numpy as np

# Example data (simulated)
A = np.random.normal(4500, 800, 1000)  # Avg 4500 RWF, std 800
B = np.random.normal(5000, 850, 1000)  # Avg 5000 RWF, std 850

# Perform two-sample t-test (A/B test)
t_stat, p_value = ttest_ind(B, A, equal_var=False)

print(f"T-statistic: {t_stat:.2f}")
print(f"P-value: {p_value:.5f}")
```

**Output Example:**

```
T-statistic: 9.45
P-value: 0.00000002
```



## **6. Interpretation**

| Metric          | Meaning                                                                               |
| --------------- | ------------------------------------------------------------------------------------- |
| **t-statistic** | How strong the difference is (signal-to-noise ratio). Large = strong difference.      |
| **p-value**     | Probability that this difference happened by chance if there were no real difference. |

### **Decision Rule:**

* **p < 0.05** → Reject (H_0): Message B **significantly increased savings**.
* **p ≥ 0.05** → Fail to reject (H_0): No strong evidence that B performs better.

---

## **7. What’s Happening “Under the Hood”**

1. The system computes the **average deposit** for A and B.
2. It looks at the **spread (variability)** of savings in both groups.
3. It measures how far apart the averages are **relative to that variability**.
4. It converts this “distance” into a **p-value** — the smaller it is, the more confident we are that the difference is real.

---

## **8. Business & Policy Implications**

| Area                      | Implication                                                                                      |
| ------------------------- | ------------------------------------------------------------------------------------------------ |
| **Product**               | Use Message B as the default → proven to drive higher deposits.                                  |
| **Finance**               | Quantifies the **ROI of behavioral messaging**.                                                  |
| **Policy Design**         | Similar experiments can be run for **subsidies, training programs, or tax awareness campaigns**. |
| **Data Science Practice** | Forms the foundation for **evidence-based decision-making**.                                     |

---

## **9. Common Mistakes to Avoid**

1. **Not randomizing users** → bias in group selection.
2. **Too small a sample size** → no power to detect real effects.
3. **Running the test too short** → early differences may vanish over time.
4. **Ignoring context** → statistical significance doesn’t always mean practical importance.

## **11. Key Takeaways for Data Scientists**

* **A/B testing = controlled two-sample test** with randomization.
* **Two-sample test = mathematical engine** inside A/B testing.
* Use it to make **evidence-based improvements** — not just guesses.
* Understand both **statistical significance (p-value)** and **practical significance (effect size)**.
* Communicate results clearly — "Message B increased savings by 11%, p < 0.001" speaks powerfully.

**Summary**

A/B testing is simply a **two-sample test with controlled randomization** — turning statistics into **actionable decisions**.

It’s how data scientists help companies, NGOs, and governments **test ideas safely, measure results, and act with confidence.**



````python
## A/B Test for different items convertion rate

from statsmodels.stats.proportion import proportions_ztest
A = [55, 88] #Purchase
B = [355, 359] #Clicks

#H0: RateA =  RateB
#H1: RateA <>  RateB

start, p_value = proportions_ztest(A, B)
print("P_value:", p_value)

# P_value = 0.002 < 0.05, reject H0, Version B's click rate significantly different from A (likely higher)
````
