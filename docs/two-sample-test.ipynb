
# **Report: Evaluating the Impact of Government Support on Rural Savings in Rwanda**

## **1. Context**

The government of Rwanda has launched a program to **support rural villagers** through financial aid, training, and access to savings mechanisms.

We want to measure:

> “Did the villagers’ **average savings increase significantly** after receiving support compared to before?”

---

## **2. Data Setup**

We have two groups:

1. **Before Support:** Savings recorded **before the program**.

   ```python
   savings_before = village_data[village_data['period']=='before']['savings']
   ```
2. **After Support:** Savings recorded **after the program**.

   ```python
   savings_after = village_data[village_data['period']=='after']['savings']
   ```

Goal: Compare these two groups statistically using a **two-sample test**.

---

## **3. Step-by-Step Logic**

1. **Calculate group averages:**

   * Mean savings before → $\bar{x}_{before}$
   * Mean savings after → $\bar{x}_{after}$

2. **Assess variability:**

   * Standard deviation shows how much savings vary within each group.

3. **Compute the difference relative to variability:**

   * The two-sample t-test calculates a **t-statistic** and **p-value**.

**Intuitive interpretation:**

* Big increase relative to variability → likely real effect
* Small increase relative to variability → could be random

---

## **4. Hypotheses**

* **Null hypothesis $(H_0)$** → The support program **did not change average savings**.
* **Alternative hypothesis $(H_a)$** → The program **increased average savings**.

---

## **5. Example Output & Interpretation**

```python
from scipy.stats import ttest_ind

t_stat, p_value = ttest_ind(savings_after, savings_before, equal_var=False)
print(t_stat, p_value)
```

| Output Example     | Meaning                                                                             |
| ------------------ | ----------------------------------------------------------------------------------- |
| `t_stat = 4.5`     | The difference in savings is **large relative to variability**                      |
| `p_value = 0.0001` | Very low probability that this difference happened by chance → program is effective |

**Decision:**

* **p < 0.05** → reject null → support program **significantly increased savings**.
* **p ≥ 0.05** → fail to reject null → no strong evidence of impact → consider program redesign.

---

## **6. Practical Implications**

1. **Policy Decisions:**

   * Confirms program effectiveness → continue or scale it.
   * Weak impact → investigate implementation gaps.

2. **Resource Allocation:**

   * Helps target funding to **villages with the most potential impact**.

3. **Future Monitoring:**

   * Provides a **baseline and statistical framework** for tracking progress over time.

4. **Communication:**

   * Simple numbers + visualization → **storytelling to stakeholders and public**.


### **Key Takeaways for Data Scientists**

* Two-sample test is **a tool to quantify impact** of interventions.
* p-value tells you **how confident you are that observed changes are real**.
* Always consider **effect size and variability**, not just averages.
