
# Reference Report: One-Sample t-Test — Foundation, Logic & Insight


##  1. Objective

**Question:**

> Is the average body mass of penguins statistically different from 4200 grams?

**Goal:**
Use statistical evidence (not guesswork) to test whether the sample mean significantly deviates from the hypothesized population mean.


##  2. Data & Context

* **Dataset:** `penguins.csv`
* **Variable of Interest:** `body_mass_g`
* **Sample:** A group of penguins with known body masses.
* **Hypothesized Population Mean (μ₀):** 4200 g


##  3. Statistical Model

### Hypotheses

| Symbol | Hypothesis Statement | Interpretation                                    |
| ------ | -------------------- | ------------------------------------------------- |
| H₀     | μ = 4200             | Penguins’ true mean body mass equals 4200 g       |
| H₁     | μ ≠ 4200             | Penguins’ true mean body mass differs from 4200 g |

### Test Type

**Two-tailed One-Sample t-Test**
We test for any difference (both higher or lower).



## 4. Core Formula



$$
t = \frac{\bar{x} - \mu_0}{s / \sqrt{n}}
$$

| Symbol      | Meaning                                                   |
| ----------- | --------------------------------------------------------- |
| $ \bar{x} $ | Sample mean                                               |
| $ \mu_0 $  | Hypothesized population mean                              |
| \( s \)       | Sample standard deviation                                 |
| \( n \)       | Sample size                                               |
| \( t \)       | Number of standard errors the sample mean is away from \( \mu_0 \) |


## 5. Decision Logic

| Step | Action                              | Outcome                                    |
| ---- | ----------------------------------- | ------------------------------------------ |
| 1    | Compute t-statistic                 | Measure deviation from μ₀                  |
| 2    | Compute p-value                     | Probability of seeing this t if H₀ is true |
| 3    | Compare p-value to α (usually 0.05) | Decide significance                        |

### Decision Rule

* If **p < 0.05** → Reject H₀ → Significant difference
* If **p ≥ 0.05** → Fail to reject H₀ → No significant difference


## 6. Python Implementation

```python
from scipy import stats
import pandas as pd

penguins = pd.read_csv("modular_program/penguins.csv")
penguins = penguins.dropna(subset=['body_mass_g'])

t_stat, p_value = stats.ttest_1samp(penguins['body_mass_g'], 4200)

print("T-statistic:", round(t_stat, 4))
print("P-value:", round(p_value, 4))
```

**Result:**

```
P-value = 0.9678
```

---

## 7. Interpretation

| Element              | Meaning                                                                    |
| -------------------- | -------------------------------------------------------------------------- |
| **P-value = 0.9678** | Very high → the data strongly supports H₀                                  |
| **Conclusion**       | Fail to reject H₀                                                          |
| **Meaning**          | The average penguin body mass is *not significantly different* from 4200 g |

## 8. Core Insights

| Category                 | Insight                                                                                                       |
| ------------------------ | ------------------------------------------------------------------------------------------------------------- |
| **Statistical Logic**    | t-test quantifies whether an observed mean is far enough from the hypothesized mean to be considered unusual. |
| **Nature of t-test**     | Used when population standard deviation is unknown. Relies on sample standard deviation.                      |
| **p-value Meaning**      | Probability of obtaining this sample (or more extreme) if H₀ were true. Not the probability that H₀ is true.  |
| **Interpretation Skill** | Failing to reject H₀ ≠ proving equality. It means there is *no sufficient evidence* to claim difference.      |
| **Sensitivity**          | t-test is sensitive to sample size — large samples detect smaller differences as significant.                 |
| **Directionality**       | One-tailed → test in one direction; Two-tailed → test in both directions.                                     |


## 9. Assumptions (Always Check Before Using t-test)

| No. | Assumption              | Description                                                | Why It Matters                         |
| --- | ----------------------- | ---------------------------------------------------------- | -------------------------------------- |
| 1   | **Random Sampling**     | Data must be randomly collected from the population.       | Prevents bias in estimation.           |
| 2   | **Independence**        | Each observation must be independent of others.            | Avoids double-counting information.    |
| 3   | **Normality**           | The data (or sample means) should be approximately normal. | Ensures validity of t-distribution.    |
| 4   | **Continuous Variable** | The tested variable must be numeric and continuous.        | Allows meaningful computation of mean. |
| 5   | **Scale Consistency**   | Same measurement units across observations.                | Maintains validity of comparisons.     |
